# **Multi-line FASTA to Single-line FASTA** **bold text**

**Read the Multi-line FASTA File  **          
We loop through the file line by line, collecting sequence lines under each header.

```python
# Initialize containers
output_lines = []
current_header = None
current_sequence = []

# Open the file
with open("multiline_input.fasta", "r") as f:
    for line in f:
        line = line.strip()
        if line.startswith(">"):
            # Save previous entry if exists
            if current_header:
                output_lines.append(current_header)
                output_lines.append(''.join(current_sequence))
            # Start new entry
            current_header = line
            current_sequence = []
        else:
            current_sequence.append(line)

    # Don't forget the last entry
    if current_header:
        output_lines.append(current_header)
        output_lines.append(''.join(current_sequence))

```

**Write to Output File **                              
We write the new two-line-per-entry format to a new file.

```python
# Write to new file
with open("singleline_output.fasta", "w") as f:
    for line in output_lines:
        f.write(line + "\n")
```

**Display First Few Lines**


```python
# Print first few lines to verify
with open("singleline_output.fasta", "r") as f:
    for _ in range(6):  # show first few lines
        print(f.readline().strip())
```
